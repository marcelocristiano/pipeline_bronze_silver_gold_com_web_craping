```python
CAMINHO_LANDINGZONE = "/Volumes/landingzone/arquivos/altura_peso_criancas"
CAMINHO_BRONZE = "/Volumes/bronze/arquivos/altura_peso_criancas"
```

```python
lsita_de_arquivos_a_processar = [arquivo.name for arquivo in dbutils.fs.ls(CAMINHO_LANDINGZONE) if arquivo.name.endswith(".csv")]
```

```python
print(lsita_de_arquivos_a_processar)
```

```python
def salva_arquivo_parquet(lista_arquivos) -> None:
    for arquivo in lsita_de_arquivos_a_processar:
        df = spark.read.csv(CAMINHO_LANDINGZONE + "/" + arquivo, header=True, inferSchema=False)
        df.write.format("parquet").save(f"{CAMINHO_BRONZE}/{arquivo}.parquet")
        print(f"Salvando arquivo {arquivo} em parquet")
```

```python
salva_arquivo_parquet(lsita_de_arquivos_a_processar)
```
