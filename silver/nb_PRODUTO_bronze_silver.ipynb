```python
from pyspark.sql.functions import *
```

```python
CAMINHO_BRONZE = "/Volumes/bronze/arquivos/produtos"
CATALOG = "silver"
SCHEMA = "vendas"
TABLE = "produtos"
```

```python
print(f"CAMINHO_BRONZE: {CAMINHO_BRONZE}")
print(f"CATALOG: {CATALOG}")
print(f"SCHEMA: {SCHEMA}")
print(f"TABLE: {TABLE}")
```

```python
# Cria schema vendas
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{SCHEMA}")
```

```python
# Ler arquivos parquet
df_produtos = spark.read.parquet(f"{CAMINHO_BRONZE}/*.parquet")
```

```python
# Altera nome de colunas e tipos
df_produtos = (
    df_produtos.withColumnRenamed("IDProduto", "id_produto") 
        .withColumnRenamed("Data", "data") 
        .withColumnRenamed("IDCliente", "id_cliente") 
        .withColumnRenamed("IDCampanha", "id_campanha") 
        .withColumnRenamed("Unidades", "unidades") 
        .withColumnRenamed("Produto", "produto") 
        .withColumnRenamed("Categoria", "categoria") 
        .withColumnRenamed("Segmento", "segmento") 
        .withColumnRenamed("IDFabricante", "id_fabricante") 
        .withColumnRenamed("Fabricante", "fabricante") 
        .withColumnRenamed("CustoUnitario", "custo_unitario") 
        .withColumnRenamed("PrecoUnitario", "preco_unitario") 
        .withColumnRenamed("CodigoPostal", "codigo_postal") 
        .withColumnRenamed("EmailNome", "email_nome") 
        .withColumnRenamed("Cidade", "cidade") 
        .withColumnRenamed("Estado", "estado") 
        .withColumnRenamed("Regiao", "regiao") 
        .withColumnRenamed("Distrito", "distrito") 
        .withColumnRenamed("Pais", "pais") 
        .withColumnRenamed("ano", "ano") 
            .withColumn("id_produto", col("id_produto").cast("int")) 
            .withColumn("data", col("data").cast("date"))
            .withColumn("id_cliente", col("id_cliente").cast("int")) 
            .withColumn("id_campanha", col("id_campanha").cast("int"))
            .withColumn("unidades", col("unidades").cast("int")) 
            .withColumn("id_fabricante", col("id_fabricante").cast("int"))
            .withColumn("custo_unitario", col("custo_unitario").cast("double")) 
            .withColumn("preco_unitario", col("preco_unitario").cast("double"))
            .withColumn("codigo_postal", col("codigo_postal").cast("int"))
            .withColumn("ano", col("ano").cast("int"))
    )
```

```python
# Salva tabela
df_produtos.write.mode("overwrite").saveAsTable(f"{CATALOG}.{SCHEMA}.{TABLE}")
```
