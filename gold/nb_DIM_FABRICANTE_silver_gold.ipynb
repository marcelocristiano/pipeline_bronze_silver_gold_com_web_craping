```python
from pyspark.sql.functions import *
```

```python
# Dados de Silver
CATALOG_SILVER = "silver"
SCHEMA_SILVER = "vendas"
TABLE_SILVER = "produtos"

# Dados de Gold
CATALOG_GOLD = "gold"
SCHEMA_GOLD = "vendas"
TABLE_GOLD = "dim_fabricantes"
```

```python
print(f"CATALOG_SILVER: {CATALOG_SILVER}")
print(f"SCHEMA_SILVER: {SCHEMA_SILVER}")
print(f"TABLE_SILVER: {TABLE_SILVER}")
print("--------------------------")
print(f"CATALOG_GOLD: {CATALOG_GOLD}")
print(f"SCHEMA_GOLD: {SCHEMA_GOLD}")
print(f"TABLE_GOLD: {TABLE_GOLD}")
```

```python
# Cria schema vendas
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG_GOLD}.{SCHEMA_GOLD}")
```

```python
df_fabricante = spark.table(f"{CATALOG_SILVER}.{SCHEMA_SILVER}.{TABLE_SILVER}").select(
    "id_fabricante",
    "fabricante",
)
df_fabricante = df_fabricante.dropDuplicates()
```

```python
df_fabricante = df_fabricante.orderBy("id_fabricante")
```

```python
df_fabricante.write.mode("overwrite").saveAsTable(f"{CATALOG_GOLD}.{SCHEMA_GOLD}.{TABLE_GOLD}")
```
